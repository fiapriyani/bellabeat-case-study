# Data Visualization

```python
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df_user_count = pd.DataFrame({
    'feature': [
        'Daily Activity',
        'Hear Rate',
        'Hourly Calories',
        'Hourly Intensities',
        'Hourly Steps',
        'Minute Calories',
        'Minute Intensities',
        'Minute METs',
        'Sleep',
        'Minute Steps',
        'Weight'
    ],
    'user_count': [
        35,
        14,
        34,
        34,
        34,
        34,
        34,
        34,
        23,
        34,
        11
    ]
})
```

```python
df_sorted = df_user_count.sort_values(
    'user_count',
    ascending=False
)

sns.barplot(
    data=df_sorted,
    x='user_count',
    y='feature',
    order=df_sorted['feature'].tolist(),
    hue='feature',
    palette='dark:b',
    legend=False
)

plt.xlim(0, 35)
plt.xlabel('Number of Users')
plt.ylabel('Feature')
plt.title('Number of Users by Feature')

plt.show()
```
